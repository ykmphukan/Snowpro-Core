# Caching

types :
- Metadata cache - resides in Cloud Service Layer
- query result cache - resides in Cloud Service Layer
- virtual warehouse data cache - resides in Query processing layer

1. Metadata Cache
 system managed cache that stores 
 - file level and table level metadata
 - resides in Cloud Service layer
 - automatically populated and maintained by snowflake

 what it includes: 
 - Micro Partition Info
 - File locations in cloud storage
 - row counts
 - null value indicators
 - col level min/max values

 how it helps?

 1. Prune micro parition - if metadata shows certain partitions don't contain relevant data, snowfalke skips reading those
 2. Avoid unneccesary I/O - Instead of scanning all files, snowfalke reads only what's necessary
 3. Speed up query planning - metadata helps snowflake optimizer make intelligent decision

 what are the benifits?

 1. Performance boost
 2. Reduce cost
 3. Zero overhead
 4. early filtering
 5. always on and automatic

2. Query Result Cache

- stores the final output of the query
- when exact query is run again, snowflake serves the result directly from the cache instead of reexecuting the query if conditions are met 
- resides in CLoud Service layer

When is it used?
- SQL text is iddentical
- user role and permision are same
- underlying data has not changed
- query ran withing 24 hrs
- not using any non deterministic functions like current_date()

What are the benefits?

- Zero compute cost
- instant response
- cost saving
- BI tool optimization
- reduced load

3. Virtual Warehouse Cache
- also known as Query Cache within virtual warehouse
- compute level, in-memory cache that stores micro partitions recently read from remote Snowflake storage (S3, Azure Blob or GCS) during query execution
- resides withing virtual warehouse, which is compute engine

when is it used?
- when query accesses table data, the warehouse detches the data from storage
- data blocks are temporarily stored in memory on the local nodes of virtual warehouse
- if a subsequent query on the same warehouse accesses overlapping data, snowfalke can reuse the in memory data, instead of refecthing from storage
- reduces I/O latency and boost performance

remember : Cache is lost if the warehouse is 
- resized
- suspendend
- restarted

What are its benefits
- Faster repeated queries for overlapping data
- reduces storage I/O cost
- boost performance
- great for batch jobs

| Feature | Metadata Cache | Query Result cache | Virtual Warehouse Cache |
|---|---|---|---|
|Location | Cloud service layer | cloud service layer | Virtual warehouse (compute layer)|
|what it stores | table/file metadata | final query results | recently accessed table data in memory |
| used when | table scan in needed | same query re run | same warehouse, same data queried again |
| valid if | table hasn't changed | query role and underlying data hasn't changed | warehouse is still running and not scaled|
| compute needed | no | no | yes |
| performance impact | prunes data scanned | skips query exection entirely | speeds up reruns, avoids I/O |
| persistance duration | varies | 24 hrs | as long as warehouse is running |
| Benefit | Reduced scan | faster result, no compute | faster query execution |
|best suited for | any query involving large tables | repeated queries from BI dashboard | heavy ETL workloads |
