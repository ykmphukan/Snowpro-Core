# System Defined Roles
- AccountAdmin: It is the top level role with control over all aspects of the account.
It inherits privileges from SecurityAdmin and SysAdmin and intended for account level tasks, not day to day object creation
- Security Admin: It can manage any object grants globally and create/manage users and roles. It is granted the global MANAGE GRANTS privlege and inherits the privleges of USERADMIN role
- SysAdmin : It has privileges to create and manage warehouse, database and all database objects(schemas,tables,views). Best practice: to have custom roles roll up to SYSADMIN role in role heirarchy.
- UserAdmin : a role dedicated solely to user and role management. It can create users and roles and manage those it owns, without ability to manage grants globally.
- OrgAdmin: a role used to manage operations at the org level, such as managing accounts. not part of accnt adm hierarchy
- Public role: role automatically granted to every user and roles .

# Securable Object Hierarchy
These are organized in a strict logical hierarchy. To access a specific object, the user must have all the necessary privileges on the object, as well as the USAGE privilege on all its parent container objects in the hierarchy.

high to low heirarchy:
Account ( warehouse ) - Database - Schema - Object

# Access Control Framework


- Discretionary Access Control (DAC) : Each object has a owner, who can in turn grant access to that object.
- Role Based Access Control (RBAC) : Access privileges are assigned to roles which are in turn assigned to users

RBAC is the administrative skeleton of snowflake. Access is granted to roles and then roles are assigned to users.

Logic: If you are in marketing team, you get the marketing role
Centralized : managed by security admin or user admin
Inheritance: Uses a hierarchy. If the Analyst role is granted to the Manager role, the manager automatically gets every permision the analyst has.

DAC is centred around ownership. Every object in Snowflake has an owner

logic: i created this table so i decide who else can see it
decentralized: the owner of an object can grant SELECT, INSERT or USAGE to any other roles
Control: If you create a table using dev_role, that role owns the table and can discretionally share it with QA role.

# DATA MASKING
Hiding the data behind the mask in real time.
Snowflake primariy uses dynamic data masking for protecting sensitive data. This feature obscures data in real time based on the user's role and access permiissions without altering the underlying data in storage.

- Row Level security : row access policies which are schema level objects that determine which rows a user can view in a table or view at query run time.
This ensures an user can see only data relevant to their roles or permissions without needing to create multiple copies of tables or complex views.

- Column level security : restricts user access to specific columns within a databse table, providing granular data protection for sensitive information like PII, financial data or health records. It ensures only autgorized users can view or modify particular fields, regardless of the apllication or query method used.
Two features :
1. Dynamic Data Masking : It is column level security feature that uses masking policies to selectively mask plain text data in table and view columns at query time
2. External Tokenization : enables account to tokenize data before loading it into SNowflake and detokenize the data at query run time. Tokenization is the process of removing sensitive data by replacing it with an undecipherable token.

# Encryption key management
Snowflake manages data encryption keys to protect customer data. This management can occur automatically without any need for customer intervention

Customers can also use the key management service in the cloud platform that hosts their snowflake account to maintain their own additional encryption key. When enabled, the combination of a snowflake maintained key and customer managed key creates a composite master key to protect customer data in snowflakeee. THis is called Tri Secret secure.

All snowflake customer data is encrypted by default. Snowfalke uses strong AES encryption with a hierarchial key model rooted in a cloud provider hosted hardware security module.

Encryption key rotation
- Keys in snowflake managed key hierarchy are automaticlly rotated by Snowflake when they are more than 30 days old
- Active keys are retired and new keys are creatd
- when snowflake determines the retired key is no longer needed, the key is destroyed.

 # Data replication and failover:
 If an entire cloud region fails, your entire data is trapped without a plan. With Snowflake data replication, you have clone data in different part of the world that is not affected due to this failure. This is called Data replication.

 Data replication is the act of copying your data from a Primary account to Secondary account. It's nt ust the tables. It's your users , roles and security policies. Snowflake always keeps the backup in sync so that it looks exactly like the original

 'Failover' : iF your main region goes down, you promote your secondary account to be the new primary. Your users are redirected and business continues without nothing happened.

 ' Failback' : once the original region is fixed, you failback. You sync your new data created during the failover and switch the primary status back to where it started.
