# WHY SNOWFLAKE?

1. Earlier companies struggled with Big Data. Databases were 
- Rigid, Expensive and didn't play well with the cloud. 

Earlier Compute and Storage were coupled together. If you need more computer power to run a report, you need to buy more
storage even if you don't need it.

```python
Snowflake solves this problem because Compute and Storage are seperate layers. You can use Snowflake's 
massive compute power to query your data and you would be charged only for the time being you use.
```

2. Earlier, different teams used the same compute resources and as a result during yearly report submit, one team(Finance) ran a massive report resulting in slowdown of another report ran by another team(Marketing).

```python
Snowflake allows different computes(Virtual Warehouses) for different teams 
so that each team can run their report without impacting the other teams.
```

3. Earlier a lot of Database administrators were required to manage a Data Warehouse to handle indexing,
partitioning.

```python
Snowflake is SaaS, doesn't require any hardware and no softwares to install and no tuning needed.
```

4. Breaking Data Silos, Secure Data Sharing, Zero Copy Cloning and Time Travel.
