# Streams

- a stream object records DML changes made to tables including inserts, updates and deletes
- a stream stores metadata about each change, so that action could be taken using metadtaa
- prcoess is called CDC ( change data capture)
- streams track all row level changes to a source table using offset but doesn't store the changed date.
- once these changes are consumed by the target table, this offset moves to the next point
- streams can be combined with tasks to set continious data pipelines
- snowpipe + stream, + task : continious data load

it maintains 3 metadata field:
- metadata$action : indicates teh dml operation recorded
- metadata$isupdate : indicates whether the operation was part of an update statement. updates to rows in the source object are represented as. apair of delete and insert records in a strea, with metdata column metadta$isupdate values set to true
- metadata$row_id : specifies the unique and immutable ID for the row, which can be used to track changes to speciifc rows over time.

# How stream works

We use merge statement for consuming the changes from stream and applying the same on Target tables.

1. To identify INSERT records
where METADATA$ACTION = 'INSERT' and METADATA$ISUPDATE='FALSE';

2. To identify UPDATE records
where METADATA$ACTION = 'UPDATE' and METADATA$ISUPDATE = 'TRUE';

3. To identify DELETE records
where METADATA$ACTION = 'DELETE' and METADATA$ISUPDATE = 'FALSE';

TYPES
1. Standard streams : it tracks all DML changes to the source object, including inserts, updates and deletes
syntax
CREATE OR REPLACE STREAM my_stream on table my_table;

2. Append only streams : It tracks row inserts only. Update and delete operations are not recorded.
CREATE OR REPLACE STREAM my_stream on table my_table
APPEND_ONLY = TRUE;

3. Insert only stream : Supported for external tables only. it tracks row inserts only.

CREATE OR REPLACE STREAM my_stream on external table my_table
INSERT_ONLY = TRUE;

# TASKS

We use tasks for scheduling in SNowflake
we can schedule
- SQL queries
- Store Procedures

Tasks can be combined with streams for implementing the continuous change data captures(CDC)
- we can maintain a DAG of tasks to keep the dependencies between tasks

Tasks require compute reourcees to execute SQL code , we can choose either
- Snowflake managed compute resources (serverless)
- User managed (virtual warehouse)

syntax

CREATE OR REPLACE TASK TASK_NAME
WAREHOUSE ='warehouse_name'
SCHEDULE = 'Time or Cron'
AFTER dep_task_name;


# SNOWFLAKE OPENFLOW
It is managed, unified data integration service designed to simplify the movement of data from virtually any source into the Snowflake AI data cloud

- Built on Apache NiFi 2.0 framework. It provides a visual low code env to uild, scale and govern complex data pipelines eliminating the need for fragmented third party ETL tools

Core Architechture

- Control Plane : Hosted within SNowsight UI. This is your mission control where you use drag and drop canvas to design, mopnitor and manage your data flows.
- Data Plane : Actual data processing happens. Snowflake offers two deployment models
1. Snowflake managed: runs within snowflake's infra( using snowpark container services)
2. Bring your own cloud(BYOC): this runtime executes within your VPC(AWS or Azure). This allows you to process sensitive data locally, complying with strict security policies while snowflake still manages the logic of the flow
