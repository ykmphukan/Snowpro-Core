# Traditional Architechture

1. Shared Disk
2. Shared Nothing

```python
Shared Disk : Single storage is being shared by multiple nodes each having their indivual compute power.
Limitations: Complexity arises when Node increases as the storage allocation reduces for each node.
```

```python
Shared Nothing: Each storage is being used by one node as allocated by resource manager.
Limitation: Hard to share information between nodes.
```

# Snowflake Architechture

Snowflake has 3 layers:
1. Cloud Services Layer ( Authentication & Access Control , Infrastructure Manager, Optimizer, Metadata Manager, Security )
2. Query Processing Layer
3. Database Storage Layer

All the benefits of Shared Disk and Shared Nothing are present.

Database Storage Layer: 
Snowflake uses a Central Data Repository for persisted data that is accesible from all compute nodes in the platform, works like a Shared Disk.

Query Processing Layer: Snowflake queries using Massive Parallel Processing compute clusters, where each node in the cluster stores a portion of the entire data set locally.

Cloud Services Layer: This is the brain of Snowflake and every management task is handled by this layer. Some of
the important things that it does are:
- Security(Compliance), Authentication & Access Control
- Snowflake Horizon Catalog
- Infrastructure Management
- Metadata Management
- Query Parsing and Optimization
