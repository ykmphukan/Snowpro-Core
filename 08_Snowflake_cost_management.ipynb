# Monitoring and Cost Management

They track credit usage by Virtual Warehouse and their supporting Cloud services.
When a threshold is hit, snowflake perform these actions:
- Notify  : Send an alert to all account admins
- Suspend : Stops the warehouse from taking new queries, but allows currently runnig queries to finish.
- Suspend immediate: Instantly cancels all running queries and shuts down the warehouse.

# Warehouse monitoring metrics
Warehouse_metering_history : 
Best for Cost, shows how many credits each warehouse cosnumed by the hour
Query_history: Best for granular performance. Shows execution time, queued time and bytes scanned for indivual queries.
Warehouse_load_history: Best for concurrency. shows if queries are running, queued or provsioning.

Importants
- reset frequency: Monitors can reset daily, weekly , monthly , yearly or never
- role required: AccountAdmin
- Cloud services : resource monitors track cloud services usage but only trigger suspension based on Warehouse
- Grace period: Credit usage resets to 0 at 12:00 AM UTC, regardless of your time zone

# Calculating Virtual Waehouse credit usage

1. per second bill rule: Snowflake bills for warehouse usage by the second, but there is a catch.
min billed: 60 sec , everytime a warehouse starts or is resized to a larger size.
2. T sized multiplier: 

| Size | Credits per hour |
| --- | ---- |
| x small | 1 |
| small | 2 |
| medium | 4 |
| large | 8 |
| extra large | 16 | 

3. Multi cluster warehouse : 
Total credits = Credits for size * no of clusters running
4. Cloud service free tier: snowflake chatrges for cloud service( like mertadata hits or security checks) only if they exceeds 10% of daily warehouse yusage

Top 5 views

- Query history : Tracks every query run in the last year. used to find expensive queries or slow performance.
- Warehouse metering history : billing view. shows credit consumption per hour per warehouse
- login_history : the security view, tracks successful and failed logins
- storage usage : shows your data footprint( database, stage and fail safe storage) for biklling
- access_history: tracks which users accessed which specific columns, crucial for data lineage
